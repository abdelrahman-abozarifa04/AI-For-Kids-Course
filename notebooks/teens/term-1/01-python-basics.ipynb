## ٥. الدوال: نجمع خطوات تحت اسم واحد

الدالة = **خوارزمية** ليها اسم. بنكتبها مرة ونستخدمها كتير.

</div>

````python
def grade(score):
    """Return a letter grade for a score out of 100."""
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 50:
        return "C"
    return "F"

for s in [95, 80, 60, 30]:
    print(s, "->", grade(s))
````

## ٦. القوائم (lists): مجموعة حاجات بالترتيب

</div>

````python
scores = [88, 92, 75, 60, 99]
print("First:", scores[0], "| Last:", scores[-1])
print("Count:", len(scores), "| Max:", max(scores), "| Average:", sum(scores) / len(scores))

scores.append(70)
scores.sort()
print("Sorted:", scores)

# list comprehension: طريقة مختصرة نعمل بيها قائمة جديدة
passed = [s for s in scores if s >= 75]
print("Passed:", passed)
````

## ٧. القواميس (dictionaries): مفتاح ← قيمة

زي القاموس بالظبط: الكلمة ومعناها. في الذكاء الاصطناعي بنستخدمها علشان نوصف حاجة بخصائصها.

</div>

````python
student = {"name": "Youssef", "age": 17, "city": "Alexandria", "skills": ["Python", "Math"]}
print(student["name"], "lives in", student["city"])

student["age"] = 18                  # نعدّل
student["grade"] = "A"               # نضيف
for key, value in student.items():
    print(f"{key}: {value}")
````

## ٩. الأخطاء (errors) وإزاي نصلحها

الغلط عادي جداً. اقرا **آخر سطر** في رسالة الخطأ: هو اللي بيقولك المشكلة.

</div>

````python
try:
    print(scores[100])
except IndexError as err:
    print("Error:", err, "-> the list is shorter than that!")

try:
    print("Age: " + 16)
except TypeError as err:
    print("Error:", err, "-> use f-strings or str(16)")
````
